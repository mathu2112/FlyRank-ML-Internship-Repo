## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

---
Signal Check 1 - CTR distribution

What I am testing: CTR relative to search position.

Why I chose this signal: CTR versus position is linked to FlyRank's CTR-fix logic. Content appearing in similar search-position ranges can have different click-through rates, so I first check whether CTR shows a clear relationship with position before using it in a rule.

What I will measure: I calculate CTR as clicks divided by impressions. I then group content into position buckets and calculate the number of observations (n) and average CTR for each bucket.

Signal Check 2 - CTR vs Search Position

Verdict: CONFIRMED. The results show a clear relationship between search position and CTR. Mean CTR decreases consistently as the average search position becomes worse, falling from 0.016720 for content in positions 1–3 to 0.000404 for content in positions 51+. The bucket counts (n) are substantial across all groups, so the pattern is supported by a large number of observations. This confirms that CTR should be interpreted together with search position rather than being used alone.

Signal Check 3 - Volume/Impressions

Verdict: MIXED
Signal: Impression volume

I grouped content into impression-volume buckets and calculated the number of observations (n) and mean CTR for each bucket. Mean CTR was similar across the groups and did not show a consistent increase or decrease as impression volume increased.

Verdict: MIXED. Impression volume alone does not show a clear relationship with CTR. However, impressions can still be useful when prioritising actions because a low-CTR item with many impressions may represent a larger potential opportunity than an item with only a few impressions.



```python
# This cell is for CODE (numbers, a query, a check).
#Importing data from huggingface
!pip install -q datasets duckdb pyarrow pandas huggingface_hub

from google.colab import userdata

HF_TOKEN = userdata.get("HF_TOKEN")

#For Accessing dataset
from datasets import load_dataset

dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    token=HF_TOKEN
)
print(dataset)

# Check some report dates
print(dataset["train"]["report_date"][:10])
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
# Look at the main performance signals
print(dataset["train"].select_columns([
    "report_date",
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position"
]).to_pandas().head())
```

```python
# ============================================================
# THREE SIGNAL TESTS — LIGHTWEIGHT VERSION
# ============================================================

import pandas as pd


# Take 1 million rows WITHOUT shuffling

train = dataset["train"]

sample_data = train.select(range(1000000))

print(sample_data)


# Convert the sample to pandas
df = sample_data.select_columns([
    "report_date",
    "content_hash_id",
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position"
]).to_pandas()


# ============================================================
# SIGNAL TEST 1 — CTR
# ============================================================

# Calculate CTR = clicks / impressions
# Rows with zero impressions will have missing CTR

df["ctr"] = df["gsc_clicks"] / df["gsc_impressions"]

df.loc[
    df["gsc_impressions"].isna() |
    (df["gsc_impressions"] == 0),
    "ctr"
] = None

print("\nSIGNAL TEST 1 — CTR SUMMARY")
display(df["ctr"].describe())


# ============================================================
# SIGNAL TEST 2 — SEARCH POSITION
# ============================================================

# Create position buckets

position_bins = [
    -float("inf"),
    3,
    10,
    20,
    50,
    float("inf")
]

position_labels = [
    "1-3",
    "4-10",
    "11-20",
    "21-50",
    "51+"
]

df["position_bucket"] = pd.cut(
    df["gsc_avg_position"],
    bins=position_bins,
    labels=position_labels
)

# Create bucket table with n and mean CTR

position_summary = (
    df
    .dropna(subset=["ctr", "position_bucket"])
    .groupby(
        "position_bucket",
        observed=True
    )
    .agg(
        n=("ctr", "count"),
        mean_ctr=("ctr", "mean")
    )
    .reset_index()
)

print("\nSIGNAL TEST 2 — CTR VS SEARCH POSITION")
display(position_summary)


# ============================================================
# SIGNAL TEST 3 — VOLUME / IMPRESSIONS
# ============================================================

# Create volume buckets

volume_bins = [
    -1,
    0,
    10,
    100,
    1000,
    float("inf")
]

volume_labels = [
    "0",
    "1-10",
    "11-100",
    "101-1000",
    "1000+"
]

df["volume_bucket"] = pd.cut(
    df["gsc_impressions"],
    bins=volume_bins,
    labels=volume_labels
)

# Create bucket table with n and mean CTR

volume_summary = (
    df
    .dropna(subset=["volume_bucket"])
    .groupby(
        "volume_bucket",
        observed=True
    )
    .agg(
        n=("ctr", "count"),
        mean_ctr=("ctr", "mean")
    )
    .reset_index()
)

print("\nSIGNAL TEST 3 — VOLUME / IMPRESSIONS")
display(volume_summary)
```

## 3. The flag-linked test

*Pick a signal one of FlyRank's real flags relies on. Does the data support the rule's assumption?*

---
Baseline rule

Based on the signal checks, I created a rule to identify content that may have a CTR improvement opportunity. The rule focuses on content with a reasonably good search position because the flag-linked test confirmed that CTR depends strongly on position. I also require a minimum number of impressions so that the item has enough search visibility to represent a meaningful opportunity.

Content is flagged when it has an average search position of 20 or better, at least 100 impressions, and a CTR below 0.5%. Flagged content receives an action score based on its impression volume and CTR gap. The reason code is LOW_CTR_GOOD_POSITION, and the action label is INVESTIGATE_CTR.

This is a hand-written baseline rule rather than a machine-learning model. The purpose is to create a transparent baseline that can later be compared with the Week 5 model.

Flag-linked test: CTR vs Search Position


I selected CTR relative to search position as my flag-linked test because search position is a signal behind FlyRank's CTR-fix logic. I calculated CTR as GSC clicks divided by GSC impressions and grouped the observations into search-position buckets. For each bucket, I calculated the number of observations (n) and mean CTR.

Verdict: CONFIRMED. Mean CTR consistently decreased as search position became worse, from 0.016720 in positions 1–3 to 0.000404 in positions 51+. This confirms that search position is an important context for interpreting CTR


```python
# ============================================================
# BASELINE ACTION RULE
# ============================================================

import numpy as np
import pandas as pd

# Create a copy for scoring
baseline_df = df.copy()

# ------------------------------------------------------------
# RULE CONDITIONS
# ------------------------------------------------------------

good_position = (
    baseline_df["gsc_avg_position"] > 0
) & (
    baseline_df["gsc_avg_position"] <= 20
)

enough_impressions = (
    baseline_df["gsc_impressions"] >= 100
)

low_ctr = (
    baseline_df["ctr"] < 0.005
)

# Content must satisfy all three conditions
baseline_df["flagged"] = (
    good_position
    & enough_impressions
    & low_ctr
)

# ------------------------------------------------------------
# CREATE ACTION SCORE
# Higher impressions + lower CTR = higher priority
# ------------------------------------------------------------

baseline_df["action_score"] = 0.0

baseline_df.loc[
    baseline_df["flagged"],
    "action_score"
] = (
    np.log1p(
        baseline_df.loc[
            baseline_df["flagged"],
            "gsc_impressions"
        ]
    )
    *
    (
        0.005
        - baseline_df.loc[
            baseline_df["flagged"],
            "ctr"
        ]
    )
)

# ------------------------------------------------------------
# REASON CODE
# ------------------------------------------------------------

baseline_df["reason_code"] = np.where(
    baseline_df["flagged"],
    "LOW_CTR_GOOD_POSITION",
    "NOT_FLAGGED"
)

# ------------------------------------------------------------
# ACTION LABEL
# ------------------------------------------------------------

baseline_df["action_label"] = np.where(
    baseline_df["flagged"],
    "INVESTIGATE_CTR",
    "NO_ACTION"
)

# ------------------------------------------------------------
# CREATE RANKED QUEUE
# ------------------------------------------------------------

ranked_queue = (
    baseline_df[
        baseline_df["flagged"]
    ]
    .sort_values(
        "action_score",
        ascending=False
    )
    [
        [
            "report_date",
            "content_hash_id",
            "gsc_impressions",
            "gsc_clicks",
            "gsc_avg_position",
            "ctr",
            "action_score",
            "reason_code",
            "action_label"
        ]
    ]
    .reset_index(drop=True)
)

# Add rank
ranked_queue.index = ranked_queue.index + 1
ranked_queue.index.name = "rank"

print("Number of items flagged:", len(ranked_queue))

display(ranked_queue.head(10))
```

---
The results show that CTR depends strongly on search position. Content appearing in lower search positions naturally receives a lower CTR, so a low CTR alone should not automatically be treated as a problem.
